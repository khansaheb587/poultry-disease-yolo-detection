```python
yaml_content = """
path: /content/drive/MyDrive/YOLO/data
train: train/images
val: val/images
test: test/images

nc: 4
names: ['cocci', 'healthy', 'ncd', 'salmo']
"""

yaml_path = '/content/drive/MyDrive/YOLO/data/data.yaml'

with open(yaml_path, 'w') as f:
    f.write(yaml_content)

print("✅ data.yaml created!")
```

```python
from ultralytics import YOLO

# YOLO11 model load
model = YOLO('yolo11m.pt')

# Training
results = model.train(
    data='/content/drive/MyDrive/YOLO/data/data.yaml',
    epochs=20,
    imgsz=640,
    batch=16,
    device=0,
    project='/content/drive/MyDrive/YOLO/output',
    name='yolo11_poultry',
    save=True,
    patience=5
)

print("✅ Training complete!")
```

```python
from ultralytics import RTDETR

# RT-DETR model load
model = RTDETR('rtdetr-l.pt')

# Training
results = model.train(
    data='/content/drive/MyDrive/YOLO/data/data.yaml',
    epochs=20,
    imgsz=640,
    batch=8,
    device=0,
    project='/content/drive/MyDrive/YOLO/output',
    name='rtdetr_poultry',
    save=True
)

print("✅ RT-DETR Training complete!")
```
